# 1. Чи відрізняється середній чек у членів програми лояльності та звичайних клієнтів?

**Гіпотеза 1**:

**Члени програми лояльності (Member) можуть мати вищий середній чек, ніж звичайні клієнти (Normal).**

Це цікаво для бізнесу, тому що дозволяє оцінити, чи справді membership стимулює клієнтів купувати більше.

````python
df.groupby('Customer type')['Total'].mean().round(2)
````

````python
df.groupby('Customer type')['Total'].agg(['mean', 'count', 'min', 'max']).round(2)
````

**Висновок**:

Бачимо, що середній чек Member вищий за Normal, тому програма лояльності потенційно може бути пов'язана з більшою сумою покупки.

**Рекомендація**:

Активніше залучати клієнтів до програми лояльності - через бонуси, персональні знижки, накопичення балів тощо.

# 2. Які категорії товарів приносять найбільший валовий дохід?

**Гіпотеза 2**:

**Не обов'язково категорія, яку купують найчастіше, приносить найбільший прибуток.**

Тут ми дивимося не просто на кількість інвойсів, а на те, які категорії генерують найбільше грошей, тому що кількість покупок і прибутковість - не одне й те саме.

````python
df.groupby('Product line')['gross income'].sum().sort_values(ascending=False).round(2)
````

Для порівняння порахуємо кількість інвойсів по категоріях:

````python
df.groupby('Product line')['Invoice ID'].count().sort_values(ascending=False)
````

````python
df['Product line'].value_counts()
````

Подивимось, яких товарів продали найбільше за кількістю одиниць:

````python
df.groupby('Product line')['Quantity'].sum().sort_values(ascending=False)
````

Наприклад:

- Food and beverages — 160 інвойсів, але в цих інвойсах могли купувати по 5–10 товарів;
- Electronic accessories — 170 інвойсів, але в кожному в середньому купували по 2 товари.

Тоді за кількістю інвойсів лідером буде Electronic accessories, а за кількістю проданих одиниць — Food and beverages.

````python
result = df.groupby('Product line').agg({
    'Invoice ID': 'count',
    'Quantity': 'sum',
    'gross income': 'sum'
})

result = result.sort_values('gross income', ascending=False)

result
````

Тут `Invoice ID` і `Quantity` означають різні речі.

Для категорії **Food and beverages**:

- **174 Invoice ID** → було 174 покупки/транзакції, у яких була ця категорія.
- **952 Quantity** → загалом у цих покупках було придбано 952 одиниці товарів цієї категорії.

Тобто в одному інвойсі покупець міг придбати не один товар, а, наприклад:

```
Invoice 1 → 3 товари Food and beverages
Invoice 2 → 5 товарів Food and beverages
Invoice 3 → 2 товари Food and beverages
```
Це означає: **174 транзакції**, у яких сумарно купили **952 одиниці товарів**.



**Висновок**

Найбільший сумарний валовий дохід приносить категорія **Food and beverages** - **2673.56**. На другому місці **Sports and travel** - **2624.90**. Найменший валовий дохід має **Health and beauty** - 2342.56.

**Рекомендації**

Катеогорії з найбільшим валовим доходом бізнес потенційно може:

- збільшувати їх асортимент
- забезпечувати достатні запаси
- активніше просувати їх
- розміщувати їх у більш помітних місцях магазину.

Можна підтримувати та розвивати категорію **Food and beverages**, оскільки вона генерує найбільший валовий дохід. Водночас варто дослідити причини низького показника **Health and beauty** - наприклад, асортимент, кількість продажів, середню ціну та кількість товарів у чеку - перш ніж вирішувати, чи потрібно змінювати її асортимент або маркетингове просування.

Варто звернтуи увагу, що це саме сумарний gross income, а не середній прибуток з одного чека. Тому для повнішого висновку варто ще подивитися кількість продажів у кожній категорії.

# 3. Чи відрізняється задоволеність клієнтів залежно від способу оплати?

**Гіпотеза 3**:

**Рейтинг задоволеності клієнтів (Rating) може відрізнятися залежно від способу оплати (Payment).**

Це вже питання не лише що і де купують, а й як купують.

````python
df.groupby('Payment')['Rating'].mean().round(2)
````

````python
df.groupby('Payment')['Rating'].agg(['count', 'mean']).round(2)
````

**Висновок**:

Середній рейтинг клієнтів практично не залежить від способу оплати. Найвищий середній рейтинг мають покупці, які платили кредитною карткою - 7.00, але різниця з готівкою (6.97) та Ewallet (6.95) дуже незначна.

Тому робити висновок, що якийсь спосіб оплати забезпечує кращий клієнтський досвід, на основі цих даних не варто.

**Рекомендація**:

Не варто робити акцент на зміні способів оплати лише для підвищення задоволеності клієнтів, оскільки середні рейтинги за трьома способами оплати майже однакові. Для покращення клієнтського досвіду доцільніше дослідити інші фактори, наприклад категорію товарів, філіал, час покупки або тип клієнта.
