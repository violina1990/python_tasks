# Тести для малих вибірок та пропорцій



**Завдання 1**. E-commerce компанія після редизайну сайту підозрює, що **середній час до покупки (time-to-purchase)** користувачів **збільшився**.
Історично середній час від першого заходу на сайт до покупки становив $\mu_0 = 123.7$ хвилин.

Команда аналітиків випадково обрала дані **7 покупців після редизайну**:
`128, 135, 121, 142, 126, 151, 123`

З допомогою наявних даних зʼясуйте, чи збільшився середній час покупки після редизайну.

Для цього виконайте наступні 6 кроків. Правильне виконання кожного з кроків оцінюється в 1 бал.

1. Запишіть нульову та альтернативну гіпотези і визначте тип тесту.
2. Обчисліть вибіркові статистики: $\bar x$, $s$, $n$.
3. Оберіть тип тесту та виконайте його (знайдіть критичне значення тесту, статистику тесту та р-значення) будь-яким пасуючим способом, щоб перевірити гіпотезу на рівні значущості $\alpha = 0.10$.
4. Прийміть рішення, чи відхиляєте ви гіпотезу $H_0$ використовуючи p-value.
5. Напишіть висновок: чи справді редизайн сайту подовжив час до покупки?
6. Чи зміниться ваше рішення при зміні рівня значущості на $\alpha = 0.05$.


## Крок 1    

- $H_0: \mu = 123.7$ - середній час до покупки не змінився
- $H_a: \mu > 123.7$ - середній час до покупки збільшився

Тип тесту: Правосторонній, бо команда підозрює, що середній час **збільшився**.

## Крок 2

 Обчислюємо вибіркові статистики: $\bar x$, $s$, $n$.



```python
import numpy as np
from scipy import stats
```

```python
# Дані з умови
times_after_redesign = np.array([128, 135, 121, 142, 126, 151, 123])
mu0 = 123.7    # середній час до редизайну
alpha = 0.10
```

```python
# Крок 2. Обчислимо вибіркову статистику
x_bar = np.mean(times_after_redesign)
s = np.std(times_after_redesign, ddof=1)   # стандартне відхилення з n-1 ступенями свободи
n = len(times_after_redesign)
```

```python
print(f"Вибіркове середнє x̄ : {x_bar:.3f}")
print(f"Вибіркове стандартне відхилення s: {s:.3f}")
print(f"Розмір вибірки n: {n}")
```

##Крок 3

Обираємо тип тесту і знаходимо:
- критичне значення тесту
- статистику тесту
- р-значення

на рівні значущості $α=0.10$.

Для вибірки розміром $n=7 < 30$ обираємо тип тесту - правосторонній **t-test**.


Чому `sf()`, а не `cdf()`?
- `stats.t.cdf(t_statistic, df)` — обчислює ймовірність ліворуч від t-статистики: $P(T \leq t)$
- `stats.t.sf(t_statistic, df)` — обчислює ймовірність праворуч від t-статистики: $P(T > t)$.

Оскільки у нас **правосторонній** t-test $(H_a: \mu > 123.7)$, p-value — це площа праворуч від обчисленої t-статистики.

```python
# Крок A. Критичне значення для правостороннього тесту
df = n - 1                             # degrees of freedom
t_crit = stats.t.ppf(1 - alpha, df)    # для правостороннього (1 - alpha)

# Крок B. Обчислимо t-статистику вручну
t_stat = (x_bar - mu0) / (s / np.sqrt(n))

# Крок C. p-value
p_value = stats.t.sf(t_stat, df)       # sf() - бо правосторонній
```

```python
print(f"Критичне значення (df={df}): {t_crit:.3f}")
print(f"t-статистика: {t_stat:.3f}")
print(f"p-value: {p_value:.4f}")
```

## Крок 4

Приймаємо рішення на основі p-value:
- p-value $≤ 0.10$ — відхиляємо нульову гіпотезу $H_0$
- p-value $> 0.10$ — не відхиляємо нульову гіпотезу $H_0$

```python
# висновок на основі р-value

if p_value < alpha:
    print(f"Відхиляємо H0: Середній час від першого заходу на сайт до покупки більший за {mu0} на рівні значущості {alpha}")
else:
    print("Не відхиляємо H0: доказів недостатньо")
```

## Крок 5

**ВИСНОВОК:** За рівня значущості $10\%$ маємо достатньо статистичних доказів того, що після редизайну сайту середній час до покупки  збільшився. Середній час у вибірці становить $132.29$ хв проти історичних $123.7$ хв.

## Крок 6

Чи зміниться ваше рішення при зміні рівня значущості на  α=0.05.

Приймаємо рішення на основі p-value:
- p-value $≤ 0.05$ — відхиляємо нульову гіпотезу $H_0$
- p-value $> 0.05$ — не відхиляємо нульову гіпотезу $H_0$

Ні, рішення не зміниться.

Маємо p-value $= 0.0420$.

За нового рівня значущості $\alpha = 0.05$:
$$
p\text{-value}=0.0420 < 0.05
$$

Отже, тут також **відхиляємо** нульову гіпотезу $H_0$. На рівні значущості $5\%$ також маємо достатньо статистичних доказів того, що середній час до покупки збільшився.

## Інший спосіб: з використанням бібліотеки SciPy



```python
# SciPy-тест — за допомогою бібліотеки SciPy
res = stats.ttest_1samp(
                        times_after_redesign,
                        popmean=mu0,
                        alternative="greater"
                        )   # _1samp - бо маємо 1 вибірку

print(f"SciPy: t = {res.statistic:.3f}, p-value = {res.pvalue:.4f}")
```

```python
res
```

```python
# висновок на основі р-value
if res.pvalue < alpha:
    print(f"Відхиляємо H0: Середній час від першого заходу на сайт до покупки більший за {mu0} на рівні значущості {alpha}")
else:
    print("Не відхиляємо H0: доказів недостатньо")
```

##**Завдання 2**

До спеціальної рекламної кампанії **23%** дорослих упізнавали логотип компанії. Після завершення кампанії відділ маркетингу провів опитування: з **1200** випадково відібраних дорослих **311** упізнали логотип.

Перевірте на рівні значущості **3%** ($\alpha=0.03$), чи дають ці дані достатні підстави стверджувати, що **тепер більше ніж 23%** дорослих упізнають логотип компанії. Для розвʼязку використовуйте бібліотеку `statsmodels`.

Зробіть висновок, чи зросла впізнаваність логотипу.

Додатково, обчисліть довірчий інтревал на заданому рівні значущості і проінтерпретуйте текстом - як він додатково пояснює прийняте нами рішення?

**Формулюємо гіпотези**

- $H_0: p = 0.23$ — частка впізнаваності не зросла
- $H_1: p > 0.23$ — частка впізнаваності зросла

Рівень значущості: $\alpha = 0.03$.

Вибірка: $n=1200$, логотип упізнали $311$ людей (кількість "успіхів").

Вибіркова частка:
$$
\hat p=\frac{311}{1200}\approx 0.2592=25.92\%
$$

```python
import statsmodels.api as sm
from statsmodels.stats.proportion import proportions_ztest, proportion_confint
```

```python
count = 311     # кількість "успіхів"
n_obs = 1200    # обсяг вибірки
p_0 = 0.23      # гіпотетична пропорція (H0)
alpha=0.03      # рівень значущості


# z-тест для однієї пропорції; prop_var=p0 задає дисперсію за H0 (стандартний підхід)
z_stat, p_value = proportions_ztest(
                                    count = count,
                                    nobs = n_obs,
                                    value = p_0,
                                    alternative = 'larger',   # правосторонній тест: p > 0.23
                                    prop_var=p_0
                                    )


print("🔹 Тест для однієї пропорції")
print(f"Z-статистика = {z_stat:.3f}")
print(f"p-value = {p_value:.4f}")
print()
if p_value < alpha:
    print(f"Оскільки  p-value={p_value:.4f} < {alpha}, відхиляємо  H0.")
    print(f"👉 Є статистично значущі докази, що частка тих, хто впізнає логотип, більша за {p_0*100}% - впізнаваність логотипу зросла.")
else:
    print("Не відхиляємо H0: підстав стверджувати зростання впізнаваності логттипу недостатньо")
```

**Висновок**: Отримані дані дають достатньо підстав стверджувати (на рівні значущості 3%), що впізнаваність логотипу компанії статистично значущо зросла і тепер становить більше ніж 23%.

```python
# Двосторонній довірчий інтервал (Сonfidence Interval)
confint = proportion_confint(
                              count,
                              nobs = n_obs,
                              alpha = alpha,
                              method = 'normal')  # без зазначеного alternative


print(f"{100*(1-alpha):.0f}% довірчий інтервал: {confint}")
```

$97\%$ довірчий інтервал становить **від $23.17\%$ до $28.66\%$**.

З надійністю $97\%$  висуваємо припущення, що істинна частка дорослих, які знають логотип компанії після рекламної кампанії, перебуває в межах від $23.17\%$ до $28.66\%$.

**Як інтервал пояснює прийняте нами рішення:** Початкове значення  $p_0 = 23\%$ знаходиться поза межами обчисленого довірчого інтервалу - воно розташоване лівіше від нижньої межі ($23.00\% < 23.17\%$). Це додатково підтверджує, що нова частка впізнаваності вища за попередню, тому відхилення $H_0$ є повністю обгрунтованим.


**Що інтервал додає до p-value**

- **Масштаб ефекту**. p-value лише каже, що зростання є. Інтервал показує, наскільки воно велике: від $≈ +0.2$ до $≈ +5.7$ відсоткових пунктів над базовими $23\%$.
- **Практична значущість**. Нижня межа дуже близька до $0.23$ (двосторонній інтервал починається з $0.2317$). Тобто зростання статистично значуще, але в «найгіршому» сценарії воно може бути майже непомітним (близько $0.2$ в.п.). Для рішення про те, чи окупилась кампанія, це суттєве застереження.
- **Точність оцінки**. Ширина інтервалу $≈ 5$ в.п. показує, що при вибірці $1200$ осіб оцінка досить груба.

**Висновок:** І тест, і довірчий інтервал на рівні значущості $\alpha = 0.03$ підтверджують, що впізнаваність логотипу зросла: справжня частка ймовірно становить $23.17\%-28.66\%$, а базове значення $23\%$ в інтервал не входить. При цьому нижня межа близька до $23\%$, тому реальне зростання може бути невеликим. Щоб його уточнити, потрібна більша вибірка.
