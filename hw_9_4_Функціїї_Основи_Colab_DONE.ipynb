4. Напишіть функцію `multipy_numbers`, яка приймає два числа як аргументи і повертає їх добуток.

  Викличіть функцію `multipy_numbers` на парах чисел
- 10 і 2
- 7 і 8
- 1024 і 999.
  
  Виведіть результат кожного виклику.
  При виконанні цього завдання з використанням циклу - отримаєте 2х балів за нього :)

Розв'язок за допомогою циклу:

````python
def multipy_numbers(a, b):
    return a * b

numbers = [(10, 2), (7, 8), (1024, 999)]

for a, b in numbers:
    print(multipy_numbers(a, b))
````

Розв'язок без циклу:

````python
a1 = 10
b1 = 2

a2 = 7
b2 = 8

a3 = 1024
b3 = 999

def multipy_numbers(a, b):
    return a * b

print(multipy_numbers(a1, b1))
print(multipy_numbers(a2, b2))
print(multipy_numbers(a3, b3))
````

5. **Ділення з перевіркою типів.**  

  1. Напишіть функцію `int_division` яка приймає на вхід два аргументи типу `int` - `a` і `b`, і виконує ділення першого аргумента на другий: `a/b`. При цьому функція перед діленням перевіряє, чи другий аргумент не дорівнює 0. Якщо другий аргумент не 0, фукнція вертає результат ділення, інакше - замість виконання ділення, вертає повідомлення на екран "Не можу виконати ділення на 0".

    Якщо типи, надані у функцію, не `int` - функція виводить повідомлення "Не працюю з такими типами даних.".

  2. Запустіть функцію на наступних парах чисел:
- 12 і 6
- 20 і 0
- 112 і 10
- 99 і 0
- 11 i 1.1

    і виведіть на екран повідомлення "Результат роботи функції: " - і далі те, що повернула функція.

  3. Дайте відповідь на питання: "Чому в деяких випадках функція повернула None?"

Розв'язок. В умові написано:

*«вертає повідомлення на екран»*

А далі:

*«виведіть ... те, що повернула функція».*

Тому краще, щоб всередині функції всюди було `return`, а `print()` був зовні:

````python
def int_division(a, b):
    if type(a) != int or type(b) != int:
        return 'Не працюю з такими типами даних.'
    elif b == 0:
        return 'Не можу виконати ділення на 0'
    else:
        return a / b

print(f'Результат роботи функції: {int_division(12, 6)}')
print(f'Результат роботи функції: {int_division(20, 0)}')
print(f'Результат роботи функції: {int_division(112, 10)}')
print(f'Результат роботи функції: {int_division(99, 0)}')
print(f'Результат роботи функції: {int_division(11, 1.1)}')
````

````python
def int_division(a, b):
    if b != 0:
        return a / b
    else:
        print('Не можу виконати ділення на 0')

print(f'Результат роботи функції: {int_division(12, 6)}')
print(f'Результат роботи функції: {int_division(20, 0)}')
````

````python
a = 12
b = 6

def int_division(a, b):
    if type(a) != int or type(b) != int:
        print('Не працюю з такими типами даних.')
    elif b == 0:
        print('Не можу виконати ділення на 0')
    else:
        return a / b  # якщо без return, то поверне None

print(f'Результат роботи функції: {int_division(a, b)}')
````

````python
def int_division(a, b):
    if type(a) != int or type(b) != int:
        print('Не працюю з такими типами даних.')
    elif b == 0:
        print('Не можу виконати ділення на 0')
    else:
        return a / b

print(f'Результат роботи функції: {int_division(12, 6)}')
print(f'Результат роботи функції: {int_division(20, 0)}')
print(f'Результат роботи функції: {int_division(112, 10)}')
print(f'Результат роботи функції: {int_division(99, 0)}')
print(f'Результат роботи функції: {int_division(11, 1.1)}')
````

В деяких випадках функція повернула `None`, оскільки в цих випадках виконувалася гілка функції, у якій не було оператора `return`.

`print()` лише виводить повідомлення на екран, але не повертає його з функції. Якщо функція не зустріла `return`, Python автоматично повертає `None`.
